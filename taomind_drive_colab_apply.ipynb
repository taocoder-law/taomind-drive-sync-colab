```python
# Cell 1 — auth + resolve bundle + upload (~5,870 files)
FOLDER_ID = '10rWSAIAdkqo5bEGkX2mG4qRcc_PheWiZ'
# Funnel on olares-1 is banned (breaks Control). Prefer a local /content upload.
BUNDLE_URL = ''  # optional override only; leave empty and upload the .tgz
MIN_BUNDLE_BYTES = 1_000  # priority pack ~few KB; full no-Work ~20MB

from google.colab import auth
auth.authenticate_user()

import hashlib, json, os, random, tarfile, tempfile, time, urllib.error, urllib.request
from pathlib import Path
from googleapiclient.discovery import build
from googleapiclient.errors import HttpError
from googleapiclient.http import MediaFileUpload

service = build('drive', 'v3', cache_discovery=False)
LIST_KW = dict(supportsAllDrives=True, includeItemsFromAllDrives=True)
WRITE_KW = dict(supportsAllDrives=True)

def _download(url: str, dest: Path, tries: int = 4) -> None:
    last = None
    for i in range(tries):
        try:
            print(f'download try {i+1}/{tries}', url)
            urllib.request.urlretrieve(url, dest)
            if dest.stat().st_size >= MIN_BUNDLE_BYTES:
                return
            last = RuntimeError(f'too small: {dest.stat().st_size}')
        except Exception as e:
            last = e
            print('download fail', type(e).__name__, e)
            time.sleep(2 ** i)
    raise RuntimeError(f'bundle download failed: {last}')

def resolve_bundle() -> Path:
    candidates = [
        Path('/content/taomind-bundle.tgz'),
        Path('/content/taomind-drive-upload-bundle-no-work.tgz'),
        Path('/content/taomind-drive-upload-bundle.tgz'),
    ]
    # Also accept anything matching under /content
    candidates += sorted(Path('/content').glob('*taomind*bundle*.tgz'))
    for p in candidates:
        if p.is_file() and p.stat().st_size >= MIN_BUNDLE_BYTES:
            print('using local bundle', p, 'bytes', p.stat().st_size)
            return p
    if not BUNDLE_URL:
        raise RuntimeError(
            'Upload taomind-drive-upload-bundle-no-work.tgz into /content/ '
            '(Funnel on olares-1 is banned; no remote download URL).'
        )
    dest = Path('/content/taomind-bundle.tgz')
    _download(BUNDLE_URL, dest)
    print('bundle_bytes', dest.stat().st_size)
    return dest

bundle = resolve_bundle()
tmpdir = Path(tempfile.mkdtemp(prefix='tm-'))
with tarfile.open(bundle, 'r:gz') as tf:
    tf.extractall(tmpdir)
src = tmpdir / 'taomind-drive-upload-bundle'
if not src.is_dir():
    # tolerate a single top-level folder rename
    kids = [p for p in tmpdir.iterdir() if p.is_dir()]
    assert len(kids) == 1, list(tmpdir.iterdir())
    src = kids[0]
print('src', src)

folder_cache = {'': FOLDER_ID}

def esc(name: str) -> str:
    return name.replace("\\", "\\\\").replace("'", "\\'")

def ensure_folder(rel_dir: str) -> str:
    rel_dir = rel_dir.strip('/')
    if rel_dir in folder_cache:
        return folder_cache[rel_dir]
    parent_rel, _, name = rel_dir.rpartition('/')
    parent_id = ensure_folder(parent_rel)
    q = (
        f"name='{esc(name)}' and '{parent_id}' in parents "
        "and mimeType='application/vnd.google-apps.folder' and trashed=false"
    )
    res = service.files().list(q=q, spaces='drive', fields='files(id,name)', pageSize=5, **LIST_KW).execute()
    files = res.get('files') or []
    if files:
        fid = files[0]['id']
    else:
        meta = {'name': name, 'mimeType': 'application/vnd.google-apps.folder', 'parents': [parent_id]}
        fid = service.files().create(body=meta, fields='id', **WRITE_KW).execute()['id']
        print('mkdir', rel_dir)
    folder_cache[rel_dir] = fid
    return fid

def find_existing(name: str, parent_id: str):
    q = (
        f"name='{esc(name)}' and '{parent_id}' in parents "
        "and trashed=false and mimeType!='application/vnd.google-apps.folder'"
    )
    res = service.files().list(q=q, spaces='drive', fields='files(id,name,md5Checksum)', pageSize=5, **LIST_KW).execute()
    return (res.get('files') or [None])[0]

def md5_file(path: Path) -> str:
    h = hashlib.md5()
    with path.open('rb') as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b''):
            h.update(chunk)
    return h.hexdigest()

def with_retry(fn, label: str, tries: int = 6):
    last = None
    for i in range(tries):
        try:
            return fn()
        except HttpError as e:
            last = e
            status = getattr(e.resp, 'status', None)
            if status in (403, 429, 500, 503) or 'rateLimitExceeded' in str(e):
                delay = min(60, (2 ** i) + random.random())
                print('retry', label, status, f'sleep {delay:.1f}s')
                time.sleep(delay)
                continue
            raise
        except Exception as e:
            last = e
            delay = min(30, (2 ** i))
            print('retry', label, type(e).__name__, f'sleep {delay}s')
            time.sleep(delay)
    raise last

files = [p for p in src.rglob('*') if p.is_file() and p.name != '_SYNC_MANIFEST.json']
files = [p for p in files if not str(p.relative_to(src)).replace('\\', '/').startswith('Work/')]
print('to_upload', len(files))
assert len(files) > 1000, 'bundle looks empty — re-download /apply.ipynb + /bundle.tgz'

ok = skipped = err = 0
t0 = time.time()
for i, p in enumerate(files, 1):
    rel = p.relative_to(src).as_posix()
    parent_rel = '/'.join(rel.split('/')[:-1])
    parent_id = with_retry(lambda: ensure_folder(parent_rel), f'mkdir:{parent_rel}')
    name = p.name
    local_md5 = md5_file(p)

    def _one():
        existing = find_existing(name, parent_id)
        if existing and existing.get('md5Checksum') == local_md5:
            return 'skip'
        media = MediaFileUpload(str(p), resumable=True)
        if existing:
            service.files().update(fileId=existing['id'], media_body=media, **WRITE_KW).execute()
        else:
            service.files().create(
                body={'name': name, 'parents': [parent_id]},
                media_body=media,
                fields='id',
                **WRITE_KW,
            ).execute()
        return 'ok'

    try:
        result = with_retry(_one, rel)
        if result == 'skip':
            skipped += 1
        else:
            ok += 1
    except Exception as e:
        err += 1
        print('FAIL', rel, type(e).__name__, e)
    if i % 100 == 0 or i == len(files):
        print(f'{i}/{len(files)} ok={ok} skipped={skipped} err={err} elapsed={time.time()-t0:.0f}s')

# Spot-check a few high-signal paths
samples = [
    'Scripts/README.md',
    'Skills/audio-narrative-reader/SKILL.md',
    'Personal/Claude-Code-Max-Access.md',
    'Summaries/1%-Lifesteal-Reading-Progress.md',
]
print('--- sample verify ---')
for rel in samples:
    lp = src / rel
    if not lp.is_file():
        print(rel, 'missing in bundle')
        continue
    parent_id = ensure_folder('/'.join(rel.split('/')[:-1]))
    existing = find_existing(lp.name, parent_id)
    want = md5_file(lp)
    got = (existing or {}).get('md5Checksum')
    print(rel, 'match' if got == want else 'MISMATCH', 'drive', got, 'local', want)

print('DONE', {'ok': ok, 'skipped': skipped, 'err': err, 'seconds': round(time.time() - t0)})
assert err == 0, f'{err} upload failures — re-run cell to retry'

```
